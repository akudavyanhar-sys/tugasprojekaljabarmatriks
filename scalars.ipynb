### Skalar


```python
# buat variabel skalar x dan simpan suatu nilai skalar
x = 7
x
```

```python
# tampilkan tipe data dari x
type(x)
```

```python
# buat variabel skalar y dan simpan suatu nilai skalar
y = 7
y
```

```python
# tampilkan tipe data dari y
type(y)
```

```python
# jumlahkan x+y
py_sum = x + y
py_sum
```

```python
# buat variabel bernama x_float dan simpan suatu nilai desimal
x_float = 7.7
x_float
```

```python
# jumlahkan x_float dengan y
float_sum = x_float + y
float_sum
```

```python
# tampilkan tipe data dari x_float
type(float_sum)
```

### Skalar menggunakan TensorFlow

```python
import tensorflow as tf
```

```python
# lakukan tahapan pada soal 1 dengan menggunakan tensorflow
x_tf = tf.Variable(7, dtype=tf.int16) # dtype is optional
x_tf
```

```python
x_tf.shape
```

```python
y_tf = tf.Variable(7, dtype=tf.int16)
```

```python
x_tf + y_tf
```

```python
tf_sum = tf.add(x_tf, y_tf)
tf_sum
```

```python
tf_sum.numpy() # note that NumPy operations automatically convert tensors to NumPy arrays, and vice versa
```

```python
type(tf_sum.numpy())
```

```python
tf_float = tf.Variable(7.7, dtype=tf.float16)
tf_float
```
