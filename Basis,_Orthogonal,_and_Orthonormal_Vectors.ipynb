### Vektor Ortogonal

```python
import numpy as np
```

```python
# buat vector i dengan elemen (1,0) dan vector j dengan elemen (0,1)
i = np.array([1, 0])
i
```

```python
j = np.array([0, 1])
j
```

```python
# buktikan bahwa vector i dan vector j itu orthogonal
np.dot(i, j)
```

### Bagaimana dengan vektor yang lain?

```python
# buat vector i dengan elemen (1,2) dan vector j dengan elemen (2, -1)
v = np.array([1, 2])
v
```

```python
w = np.array([2, -1])
w
```

```python
# Bagaimana dengan v=(1,2) dan w=(2,-1), apakah juga orthogonal?
np.dot(v, w)
```
